# Challenge 4:

Write the python code  to find the factorial of a number using a loop. Thefactorial of a number (integer positive) is the sum of multiplication of all the integers smaller than that positive integer. For example, factorial of **5** is:

5 * 4 * 3 * 2 * 1 which equals to 120. 

How you can do that in a python script?, please try your best without using google.

```python
# can write factorial of number n as n! = 1 * 2 * 3 * ... (n-1) * n
# So want to tackle this using a loop - where each loop multiplies the next number onto the existing sum (starting with 1)

#Choosw random number - try a small number first for simplicity
n = 5

# Set the first number to equal 1 (not 0 - as then the sum will always 0)
n_factorial = 1
for i in range(1,n+1):
    n_factorial = i*n_factorial #keep multiplying in the next consecutive number to the running total
    
print(f"The factorial of {n} is {n_factorial}")
```

## Description: 
Summary: Used a for loop to calculate the factorial of 5. Did this by defining the initial factorial value to be 1, so then the loop multiplies progressively onto this value then printing the final value using an f-string again for practice.

* Go to https://www.w3schools.com/python/exercise.asp?filename=exercise_functions1 and complete the 6 excersices. Functions are a very important part of Python so we will use extra time to practice that.

```python
#Declare variable for temp in Fahrenheit - but needs to be a float not an integer just in case
temp_f = float(input("Temperature in Fahrenheit = "))

# Define a function to convert the temp in Fahrenheit to Celsius 
# Found the general conversion equation just on google
def conv_temp_f_to_c(temp_f):
    return ((temp_f - 32)*5/9)

# Using the function:
temp_f_converted = conv_temp_f_to_c(temp_f)

# Printing the output of the function - with a limit of 1 d.p. since its possible to get a recurring decimal through division here
print(f"Temperature converted to Celsius = {temp_f_converted:.1f}")

# Note to say that the recommended W3schools exercixes have been complete
print("\n")
print("W3 Schools Excercises have been completed") 
```

# Challenge 8:

You have the following dataframes 'sales_data' and 'movie_ratings'

```python
import pandas as pd

sales_data_dict = {
    'Date': ['2022-01-01', '2022-01-02', '2022-01-03', '2022-01-04'],
    'Region': ['North', 'South', 'East', 'West'],
    'Product': ['A', 'B', 'A', 'C'],
    'Quantity': [15, 8, 12, 10],
    'Revenue': [1500, 1200, 1800, 900]
}

sales_data = pd.DataFrame(sales_data_dict)


movie_ratings_dict = {
    'Title': ['Movie1', 'Movie2', 'Movie3', 'Movie4'],
    'Genre': ['Action', 'Drama', 'Comedy', 'Thriller'],
    'Year': [2015, 2010, 2018, 2011],
    'Rating': [8.5, 7.8, 9.2, 8.0]
}

movie_ratings = pd.DataFrame(movie_ratings_dict)

print(sales_data)
print("\n")
print(movie_ratings)
```

```python
import pandas as pd

#1) To display first five rows use head() or head(5)
# But this doesn't really matter here since the sales_data df only has 4 row entries anyway - but keeping the mindset that this data may be updated use head(5)
print("1) The first 5 rows of the DataFrame 'sales_data' is:")
print(sales_data.head())

print() # I'm using "\n" to make it a bit easier to read

#2) Filter the dataframe using loc to display rows where 'Quantity'>10 and 'Region' == 'North'
# Asking for AND not OR so should only display the rows where both conditions are True: checked using https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.loc.html
print("2) Rows in which 'Quantity'> 10 and the 'Region' is 'North':" )
print(sales_data.loc[(sales_data['Quantity']>10)  & (sales_data['Region']=='North')])

print()

#3) Using iloc to display values in first row and first three columns
print("3) Values in the first row and first three columns")
print(sales_data.iloc[0,:3]) #checked using:https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.iloc.html

print()

#4) Calculating the total sales for each region:
#Note that the total sales per region is already summarised due to lack of other entries so could just do: print(sales_data[['Region', 'Revenue']])
#But working with the assumption that the data could be added to/changed I will include my process of how i would go about calulating this if there were more entries:
#I would use groupby to group the data by region then use sum for the revenue column - GG3209 and https://www.geeksforgeeks.org/python/pandas-groupby-and-sum/
print("4) The total sales for each region are:")
print(sales_data.groupby("Region")["Revenue"].sum())
print()

#5) To display the last three rows of the df movie_ratings
# use tail(3) - learnt in GG3209
print("5)The last 3 rows in movie_ratings are:")
print(movie_ratings.tail(3))
print()

#6) The display the average rating for each genre
#Again this is actually already summarised in the df due to small number of entries so could just do: print(movie_ratings[['Genre','Rating']])
#Assuming that more entries will be added i would use groupby and sum like for 4):
print("6)The average rating for each genre is:")
print(movie_ratings.groupby("Genre")["Rating"].mean())
```

```python
raster_path = 'data/elev.tif'
raster = rasterio.open(raster_path)
show(raster)
```

```python
#To see what the df looks like statistically:
print(gdf.describe())

# Then want a list of the column names to pick one for the map:
print(list(gdf))
```

```python
# Now create explore choropleth map using only one column e.g. speed limit of the recorded accident locations
# followed the link and explored with some other parameters - i tried scheme = 'Quantiles' too but found that there aren't enough unique values in array to form 5 classes so stayed with EqualInterval 
gdf.explore('SPEED_LIMI', cmap = "Reds",legend = True, scheme = 'EqualInterval')
```

**Figure**: Recorded speed limits at road accident locations

```python
#Creating the histogram plot using the link as guidance
#After copy-paste testing the code from the website i wanted to make a few changes to my plot - inlcuding changing the axis labels to something a bit more descriptive
#To alter the axis labels i found that i couldn't do it through rasterio but instead pass axes to the ax argument and after set axes.set_xlabel and axes.set_ylabel (found on:https://stackoverflow.com/questions/70627355/rasterio-plot-show-hist-how-to-change-modify-x-label-y-label) 
from rasterio.plot import show_hist
import matplotlib.pyplot as plt

#Creating a Matplot figure so that the histogram and axes can show on the same graph:
fig, ax = plt.subplots()

show_hist(
    (raster,1), #tried using (raster) like the multiband histogram example from the website but the legend displayed a generator-object reference not the band number so went back and changed it after reviewing the website again for plotting singleband rasters with (raster, 1) for band 1 specifically
    bins=50, lw=0.0,stacked=False,  #50 bins gives good detail, used lw=0.0 like the example to better appearance
    alpha=0.8, masked=True,  #chose masked=True to exclude any NoData values, chnaged alpha from 0.3 to 0.8 as changing raster -> (raster,1) switches the hist colour from pink to yellow and became a bit hard to see at 0.3 value 
    histtype='stepfilled', title="Elevation raster histogram", #kept stepfilled as histtype as recommended and changed title to something with context
    ax=ax) #needed to add ax=ax to ensure that the histogram was plotted onto the specified axes

#I felt like i should add in some axis labels which explains what the histogram is showing for better readability 
ax.set_xlabel("Elevation value") #To replace DN (digital number) since given the context that the input raster is an elevation tiff i would assume dn represents elevation in metres
ax.set_ylabel("Frequecy of pixels") #more descriptive than just frequency - the y-axis is showing how many pixels are within each height (m)
plt.show() #to show the complete histogram with the new axes
```
